```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from rdkit import Chem
from rdkit.Chem import AllChem

# Diccionario de SMILES CORREGIDOS para aminoácidos (formato estándar)
aa_smiles = {
    'M': 'CSCC(C(=O)O)N',  # Metionina (M) - Corregido
    'V': 'CC(C)C(C(=O)O)N',  # Valina (V) - Corregido
    'T': 'CC(O)C(C(=O)O)N',  # Treonina (T) - Corregido
}

# Secuencia de ejemplo: "MVT"
secuencia_proteina = "MVT"
mols = []
legends = []

# Crear moléculas RDKit y calcular coordenadas 3D
for aa in secuencia_proteina:
    if aa in aa_smiles:
        mol = Chem.AddHs(Chem.MolFromSmiles(aa_smiles[aa]))
        AllChem.EmbedMolecule(mol, AllChem.ETKDG())
        AllChem.UFFOptimizeMolecule(mol)
        mols.append(mol)
        legends.append(f"{aa}")

# Dibujar cada estructura en 3D usando matplotlib
if mols:
    fig = plt.figure(figsize=(12, 4))
    for idx, mol in enumerate(mols):
        ax = fig.add_subplot(1, len(mols), idx + 1, projection='3d')
        conf = mol.GetConformer()
        
        # Graficar átomos
        xs, ys, zs = [], [], []
        for atom in mol.GetAtoms():
            pos = conf.GetAtomPosition(atom.GetIdx())
            xs.append(pos.x)
            ys.append(pos.y)
            zs.append(pos.z)
        ax.scatter(xs, ys, zs, s=50, c='skyblue', edgecolors='k', depthshade=True)
        
        # Graficar enlaces
        for bond in mol.GetBonds():
            start = bond.GetBeginAtomIdx()
            end = bond.GetEndAtomIdx()
            pos1 = conf.GetAtomPosition(start)
            pos2 = conf.GetAtomPosition(end)
            ax.plot([pos1.x, pos2.x], [pos1.y, pos2.y], [pos1.z, pos2.z], c='gray')
        
        ax.set_title(f"Estructura 3D: {legends[idx]}")
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_zticks([])
        ax.set_box_aspect([1, 1, 1])  # Para escalas uniformes
        
    plt.tight_layout()
    plt.suptitle("Aminoácidos en 3D para la Secuencia 'MVT'", y=1.05)
    plt.show()
else:
    print("No se pudieron generar las estructuras 3D. Revisa los SMILES.")
```
