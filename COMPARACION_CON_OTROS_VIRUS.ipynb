```python
from Bio import SeqIO
from Bio import pairwise2
from Bio.pairwise2 import format_alignment

# Cargar todas las secuencias descargadas
secuencias = list(SeqIO.parse("parvoviridae_database.fasta", "fasta"))

# Elegimos la primera como la referencia (puedes cambiar el índice si tienes otra en mente)
ref_seq = secuencias[0]

print(f"Comparando: {ref_seq.id} ({len(ref_seq.seq)} bp)\n")

# Comparar con las demás
for i, target_seq in enumerate(secuencias[1:], start=1):
    alignments = pairwise2.align.globalxx(ref_seq.seq, target_seq.seq, one_alignment_only=True)
    score = alignments[0].score
    length = max(len(ref_seq.seq), len(target_seq.seq))
    identidad = (score / length) * 100

    print(f"🔹 Comparación con {target_seq.id}")
    print(f"   Identidad estimada: {identidad:.2f}%\n")
```
