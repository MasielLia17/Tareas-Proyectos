```python
#TAREA NRO.5
#Estudiante: Masiel Aguilar                  Código: 87770

# Ejercicio 4-7: Concentración molar de K⁺ en solución de K₃Fe(CN)₆

import pubchempy as pcp
from rdkit import Chem

# Parámetros de entrada
compound = "K3Fe(CN)6"
concentration_ppm = 63.3  # ppm
molecular_weight = 329.3  # g/mol
density = 1.00  # g/mL (asumido para solución diluida)
ion_formula = "K⁺"
ions_per_molecule = 3  # K3Fe(CN)6 → 3 K⁺ + Fe(CN)6^-3

print(f"Compuesto: {compound}")
print(f"Concentración: {concentration_ppm} ppm")
print(f"Peso molecular: {molecular_weight} g/mol")
print(f"Densidad de la solución: {density} g/mL")
print(f"Ion de interés: {ion_formula}")
print(f"Iones {ion_formula} por molécula: {ions_per_molecule}")
print()

# SOLUCIÓN PASO A PASO
print("SOLUCIÓN:")
print("-" * 40)

# Paso 1: Conversión ppm a mg/L
print("Paso 1: Conversión de ppm a mg/L")
print("Para soluciones diluidas, ppm ≈ mg/L")
mg_per_L = concentration_ppm
print(f"{concentration_ppm} ppm = {mg_per_L} mg {compound}/L")
print()

# Paso 2: Conversión mg/L a g/L
print("Paso 2: Conversión de mg/L a g/L")
g_per_L = mg_per_L / 1000
print(f"{mg_per_L} mg/L ÷ 1000 = {g_per_L:.4f} g/L")
print()

# Paso 3: Cálculo de la molaridad del compuesto
print("Paso 3: Cálculo de la molaridad del compuesto")
molarity_compound = g_per_L / molecular_weight
print(f"Molaridad = {g_per_L:.4f} g/L ÷ {molecular_weight} g/mol")
print(f"Molaridad de {compound} = {molarity_compound:.6f} M")
print(f"Molaridad de {compound} = {molarity_compound:.3e} M")
print()

# Paso 4: Cálculo de la concentración del ion K⁺
print("Paso 4: Cálculo de la concentración del ion K⁺")
print(f"Cada molécula de {compound} produce {ions_per_molecule} iones {ion_formula}")
ion_concentration = molarity_compound * ions_per_molecule
print(f"[{ion_formula}] = {molarity_compound:.3e} M × {ions_per_molecule}")
print(f"[{ion_formula}] = {ion_concentration:.6f} M")
print(f"[{ion_formula}] = {ion_concentration:.3e} M")
print()

# RESULTADO FINAL
print("="*60)
print("RESULTADO FINAL:")
print(f"La concentración molar de {ion_formula} = {ion_concentration:.2e} M")
print("="*60)
print()

# Análisis dimensional
print("ANÁLISIS DIMENSIONAL:")
print("-" * 25)
print("ppm × (g/1000mg) × (mol/g) × (iones/mol) = mol iones/L = M")
print(f"{concentration_ppm} × (1/1000) × (1/{molecular_weight}) × {ions_per_molecule}")
print(f"= {ion_concentration:.3e} M")
print()

# Función para resolver ejercicios similares
def concentracion_ionica(ppm, peso_molecular, iones_por_mol, nombre_compuesto="", nombre_ion=""):
    """
    Calcula la concentración iónica a partir de ppm del compuesto
    
    Parámetros:
    - ppm: concentración en partes por millón
    - peso_molecular: peso molecular del compuesto en g/mol
    - iones_por_mol: número de iones específicos por molécula
    - nombre_compuesto: nombre del compuesto (opcional)
    - nombre_ion: nombre del ion (opcional)
    """
    g_por_L = ppm / 1000
    molaridad_compuesto = g_por_L / peso_molecular
    concentracion_ion = molaridad_compuesto * iones_por_mol
    
    print(f"Compuesto: {nombre_compuesto if nombre_compuesto else 'No especificado'}")
    print(f"Ion: {nombre_ion if nombre_ion else 'No especificado'}")
    print(f"Concentración: {ppm} ppm → {concentracion_ion:.3e} M")
    
    return concentracion_ion

print("FUNCIÓN PARA OTROS EJERCICIOS:")
print("-" * 35)

# Ejemplo con el mismo ejercicio
resultado_funcion = concentracion_ionica(63.3, 329.3, 3, "K₃Fe(CN)₆", "K⁺")
print(f"Verificación con función: {resultado_funcion:.3e} M")
print()

# Ejemplos adicionales
print("EJEMPLOS CON OTROS COMPUESTOS:")
print("-" * 35)

# NaCl → Na⁺
print("Ejemplo 1: 100 ppm de NaCl (PM = 58.44 g/mol) → Na⁺")
conc_na = concentracion_ionica(100, 58.44, 1, "NaCl", "Na⁺")
print()

# CaCl₂ → Ca²⁺
print("Ejemplo 2: 200 ppm de CaCl₂ (PM = 110.98 g/mol) → Ca²⁺")
conc_ca = concentracion_ionica(200, 110.98, 1, "CaCl₂", "Ca²⁺")
print()

# CaCl₂ → Cl⁻
print("Ejemplo 3: 200 ppm de CaCl₂ (PM = 110.98 g/mol) → Cl⁻")
conc_cl = concentracion_ionica(200, 110.98, 2, "CaCl₂", "Cl⁻")
print()

print("="*60)
print("RESUMEN FINAL DEL EJERCICIO 4-7:")
print(f"Concentración de K⁺ = {ion_concentration:.2e} M = {ion_concentration*1000:.3f} mM")
print("="*60)
```
