```python
from docx import Document
from docx.shared import Inches, Pt
from docx.oxml.ns import qn
from docx.enum.section import WD_HEADER_FOOTER
from docx.oxml import OxmlElement
from PIL import Image
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT
from docx.shared import RGBColor
from docx.shared import Inches
from docx.shared import Cm
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT
doc = Document()

# 1. Encabezado de la empresa
company_para = doc.add_paragraph()
company_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
company_para.paragraph_format.line_spacing = 1.15
company_para.paragraph_format.space_after = Pt(0)
company_run = company_para.add_run("Company Name Here")
company_run.bold = True
company_run.font.size = Pt(36)
company_run.font.name = 'Times New Roman'
# Forzar Times New Roman en caso de que no se aplique correctamente
company_run._element.rPr.rFonts.set(qn('w:eastAsia'), 'Times New Roman')

# 2. Dirección
address_para = doc.add_paragraph()
address_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
address_para.paragraph_format.line_spacing = 1.15
address_para.paragraph_format.space_after = Pt(0)
address_para.add_run("[Street Address], [City, ST ZIP Code]")

# 3. Teléfono
contact_para = doc.add_paragraph()
contact_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
contact_para.paragraph_format.line_spacing = 1.15
contact_para.paragraph_format.space_after = Pt(0)
contact_para.add_run("[Phone: 555-555-55555] [Fax: 123-123-123456]")

# 4. Email
email_para = doc.add_paragraph()
email_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
email_para.paragraph_format.line_spacing = 1.15
email_para.paragraph_format.space_after = Pt(0)
email_para.add_run("[abc@example.com]")

# 5. Título "Contract of Sale"
title_para = doc.add_paragraph()
title_para.alignment = WD_ALIGN_PARAGRAPH.CENTER
title_para.paragraph_format.line_spacing = 1.15
title_para.paragraph_format.space_after = Pt(0)
title_run = title_para.add_run("Contract of Sale")
title_run.bold = True
title_run.font.name = 'Arial'
title_run.font.size = Pt(37)

# Añadir espacio después del título
doc.add_paragraph()

# 6. Texto del acuerdo
agreement_para = doc.add_paragraph()
agreement_text = "THIS AGREEMENT made this _____________ day of _____________, 20___, by and between _____________ (First Party) and _____________ (Second Party)."
agreement_para.add_run(agreement_text)

# 7. WITNESSETH
witness_para = doc.add_paragraph()
witness_text = "WITNESSETH: That in consideration of the mutual covenants and agreements to be kept and performed on the part of said parties hereto, respectively as herein stated:"
witness_para.add_run(witness_text)

# 8. Cláusulas numeradas
# Primera cláusula
clause1_para = doc.add_paragraph()
clause1_para.add_run("I.\tSaid party of the first part covenants and agrees that it shall:")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")

# Segunda cláusula
clause2_para = doc.add_paragraph()
clause2_para.add_run("II.\tAnd said party of the second part covenants and agrees that it shall:")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")

# Tercera cláusula
clause3_para = doc.add_paragraph()
clause3_para.add_run("III.\tOther terms to be observed by and between the parties:")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")
doc.add_paragraph("\t_________________________________________________")

# 9. Forma de pago
payment_para = doc.add_paragraph()
payment_run = payment_para.add_run("Form of payment")
payment_run.underline = True

# Lista de formas de pago
payment_list = [
    "o\tPrepayment,",
    "o\topen account with remittance at specified date",
    "o\tletter of credit,",
    "o\tpayment by compensated acceptable third party such as parent company or guarantor."
]

for payment_item in payment_list:
    payment_item_para = doc.add_paragraph()
    payment_item_para.add_run(payment_item)

doc.add_paragraph()  # Espacio

# 10. Tabla de Purchaser Details
purchaser_table = doc.add_table(rows=1, cols=1)
purchaser_table.alignment = WD_TABLE_ALIGNMENT.CENTER
purchaser_cell = purchaser_table.cell(0, 0)
purchaser_cell.text = "Purchaser Details"
purchaser_cell.paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER

# Tabla de información del comprador
buyer_table = doc.add_table(rows=4, cols=2)
buyer_table.style = 'Table Grid'

# Fila Name
buyer_table.cell(0, 0).text = "Name"
buyer_table.cell(0, 1).text = ""

# Fila Address
buyer_table.cell(1, 0).text = "Address"
buyer_table.cell(1, 1).text = ""

# Fila Telephone/Email
buyer_table.cell(2, 0).text = "Telephone"
buyer_table.cell(2, 1).text = "Email"

# Fila Recipients
buyer_table.cell(3, 0).text = "Recipient(s): List all recipients and contact information below."
# Unir las celdas de la fila de recipients
buyer_table.cell(3, 0).merge(buyer_table.cell(3, 1))

# 11. Tabla de Recipients
recipients_table = doc.add_table(rows=5, cols=3)
recipients_table.style = 'Table Grid'

# Headers
recipients_table.cell(0, 0).text = "Name"
recipients_table.cell(0, 1).text = "Contact Information"
recipients_table.cell(0, 2).text = "Relationship to Recipient"

# Filas vacías para completar
for i in range(1, 5):
    recipients_table.cell(i, 0).text = ""
    recipients_table.cell(i, 1).text = ""
    recipients_table.cell(i, 2).text = ""

doc.add_paragraph()  # Espacio

# 12. Interment Rights Details
interment_table = doc.add_table(rows=1, cols=1)
interment_table.alignment = WD_TABLE_ALIGNMENT.CENTER
interment_cell = interment_table.cell(0, 0)
interment_cell.text = "Interment Rights Details"
interment_cell.paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER

# Tabla de Cemetery
cemetery_table = doc.add_table(rows=4, cols=4)
cemetery_table.style = 'Table Grid'

# Primera fila - Cemetery
cemetery_table.cell(0, 0).text = "Cemetery"
# Unir las tres celdas restantes de la primera fila
cemetery_table.cell(0, 0).merge(cemetery_table.cell(0, 3))

# Segunda fila - Headers
cemetery_table.cell(1, 0).text = "No. of Graves"
cemetery_table.cell(1, 1).text = "Section"
cemetery_table.cell(1, 2).text = "Plot"
cemetery_table.cell(1, 3).text = "Quadrant(s)"

# Tercera fila - vacía para completar
for j in range(4):
    cemetery_table.cell(2, j).text = ""

# Cuarta fila - Notes
cemetery_table.cell(3, 0).text = "Notes"
# Unir las tres celdas restantes
cemetery_table.cell(3, 0).merge(cemetery_table.cell(3, 3))

doc.add_paragraph()  # Espacio

# 13. Deceased Information
deceased_table = doc.add_table(rows=1, cols=1)
deceased_table.alignment = WD_TABLE_ALIGNMENT.CENTER
deceased_cell = deceased_table.cell(0, 0)
deceased_cell.text = "Deceased Information (if applicable)"
deceased_cell.paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER

# Tabla de información del fallecido
deceased_info_table = doc.add_table(rows=3, cols=3)
deceased_info_table.style = 'Table Grid'

# Primera fila - Name
deceased_info_table.cell(0, 0).text = "Name"
# Unir las dos celdas restantes
deceased_info_table.cell(0, 0).merge(deceased_info_table.cell(0, 2))

# Segunda fila - Place and Date of Death
deceased_info_table.cell(1, 0).text = "Place of Death"
deceased_info_table.cell(1, 1).text = ""
deceased_info_table.cell(1, 2).text = "Date of Death"

# Tercera fila - vacía para completar
for k in range(3):
    deceased_info_table.cell(2, k).text = ""

# Guardar el documento
print("Documento 'Contract_of_Sale.docx' generado exitosamente.")
doc.save('Contract_of_Sale.docx')
```
