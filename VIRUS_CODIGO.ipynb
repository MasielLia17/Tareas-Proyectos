```python
from Bio import Entrez, SeqIO

# Tu correo electrónico (requerido por NCBI)
Entrez.email = "tucorreo@example.com"

# Buscar genomas completos de virus en la familia Parvoviridae
search_term = 'Parvoviridae[Organism] AND complete genome[Title]'

# Realizar búsqueda en NCBI
handle = Entrez.esearch(db="nucleotide", term=search_term, retmax=20)
record = Entrez.read(handle)
handle.close()

# Extraer los IDs de los genomas encontrados
ids = record['IdList']
print(f"Se encontraron {len(ids)} genomas.")

# Descargar y guardar las secuencias en archivo FASTA
with open("parvoviridae_database.fasta", "w") as output_handle:
    for id in ids:
        handle = Entrez.efetch(db="nucleotide", id=id, rettype="fasta", retmode="text")
        seq = handle.read()
        handle.close()
        output_handle.write(seq)

print("Base de datos FASTA creada: parvoviridae_database.fasta")

```
